# Gene Expression Analysis
Calculate mean expression and plot the 10 highest expressed genes.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv("gene_expression.csv")

# Get all sample columns (everything except the gene name)
sample_cols = [c for c in df.columns if c != "gene symbol"]

# Calculate mean expression across all samples for each gene
df["mean_expression"] = df[sample_cols].mean(axis=1)

# Sort by highest mean expression and take top 10
top10 = df.sort_values("mean_expression", ascending=False)[
    ["gene symbol", "mean_expression"]
].head(10)

print(top10)

```

```python
# Plot top 10 as horizontal bar chart
top10.plot(kind='barh', x='gene symbol', y='mean_expression', legend=False);
plt.title('Top 10 Expressed Genes');
plt.xlabel('Mean Expression');
plt.ylabel('Gene Symbol');
plt.tight_layout();
plt.savefig('top_genes.png')
```
