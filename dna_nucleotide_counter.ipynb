## DNA Nucleotide Counter

```python
dna = 'ATGCGATACGCTTACGATCGATCGCTAGCTAGCTAGCTGATCGA'
print(f"DNA Sequence: {dna}")
```

### Count each nucleotide and print results

```python
count_A = dna.count('A')
count_T = dna.count('T')
count_G = dna.count('G')
count_C = dna.count('C')

print(f'Count A: {count_A}')
print(f'Count T: {count_T}')
print(f'Count G: {count_G}')
print(f'Count C: {count_C}')
```

### Calculate GC content

```python
gc_content = (count_G + count_C) / len(dna) * 100
print(f'GC Content: {gc_content:.2f}%')
```

```python
counts_dict = {base: dna.count(base) for base in 'ATGC'}
print(f'Nucleotide Counts (Dictionary): {counts_dict}')
```

```python
import matplotlib.pyplot as plt

nucleotides = list(counts_dict.keys())
counts = list(counts_dict.values())

plt.figure(figsize=(8, 6))
plt.bar(nucleotides, counts, color=['blue', 'green', 'red', 'purple'])
plt.xlabel('Nucleotide')
plt.ylabel('Count')
plt.title('Distribution of Nucleotides in DNA Sequence')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
