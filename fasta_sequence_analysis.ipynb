# FASTA Sequence Analysis
Parse a local FASTA file and summarize its DNA sequences.

```python
# Import necessary modules
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
import pandas as pd
```

```python
# Define the FASTA file path relative to this notebook.
fasta_file_path = "sequence.fasta"

# Parse the FASTA file.
records = list(SeqIO.parse(fasta_file_path, "fasta"))

print(f"Found {len(records)} sequences in '{fasta_file_path}'.\n")

# Process and print details for each record.
for index, record in enumerate(records, start=1):
    print(f"--- Sequence {index} ---")
    print(f"ID: {record.id}")
    print(f"Description: {record.description}")
    print(f"Length: {len(record.seq)}")
    print(f"GC: {gc_fraction(record.seq) * 100:.2f}%")
    print(f"Sequence: {str(record.seq[:50])}...")
    print()
```

```python
# Create a summary table if there are multiple sequences
if len(records) > 1:
    summary_data = []
    for r in records:
        summary_data.append({'ID': r.id, 'Length': len(r.seq)})

    summary_df = pd.DataFrame(summary_data)
    print("Summary Table of Sequences:")
    display(summary_df)
else:
    print("Only one sequence found, no summary table needed.")
```
